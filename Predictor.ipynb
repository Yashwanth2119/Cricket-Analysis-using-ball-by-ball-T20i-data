```python
# Compute Overs for each Match innings (vectorized)
t20['overs'] = t20['Over'].astype(str) + '.' + t20['Ball'].astype(str)
t20['overs'] = pd.to_numeric(t20['overs'], errors='coerce') - 1

# Use group-wise shift to compute last-5-overs runs/wickets efficiently
# There are 6 balls per over; 5 overs -> 30 valid deliveries. We shift by 30 rows within each Match ID and Innings group.
t20.sort_values(['Match ID','Innings','Over','Ball'], inplace=True)
grp = t20.groupby(['Match ID','Innings'])
t20['runs_last_5_overs'] = t20['Innings Runs'] - grp['Innings Runs'].shift(30).fillna(0)
t20['wickets_last_5_overs'] = t20['Innings Wickets'] - grp['Innings Wickets'].shift(30).fillna(0)

# For very early balls where shift produced full Innings Runs, ensure values are non-negative
t20['runs_last_5_overs'] = t20['runs_last_5_overs'].clip(lower=0)
t20['wickets_last_5_overs'] = t20['wickets_last_5_overs'].clip(lower=0)

t20
```

```python
irrelevant = ['Unnamed: 0', 'Match ID', 'Date', 'Venue','Innings', 'Batter', 'Non Striker', 'Bowler', 'Over', 'Ball',
       'Batter Runs', 'Extra Runs', 'Runs From Ball', 'Ball Rebowled',
       'Extra Type', 'Wicket', 'Method', 'Player Out', 'Runs to Get', 'Balls Remaining',
       'Winner', 'Chased Successfully', 'Total Batter Runs',
       'Total Non Striker Runs', 'Batter Balls Faced',
       'Non Striker Balls Faced', 'Player Out Runs', 'Player Out Balls Faced',
       'Bowler Runs Conceded', 'Valid Ball']
print(f'Before Removing Irrelevant Columns : {t20.shape}')
t20 = t20.drop(irrelevant, axis=1) # Drop Irrelevant Columns
print(f'After Removing Irrelevant Columns : {t20.shape}')
t20.head(124)
```

```python
const_teams = ['Afghanistan', 'Australia', 'Bangladesh',
               'England', 'India', 'Ireland', 'New Zealand', 'Pakistan',
               'South Africa', 'Sri Lanka', 'West Indies', 'Zimbabwe']
```

```python
print(f'Before Removing Inconsistent Teams : {t20.shape}')

# Remove the Non-Consistent Teams
t20 = t20[(t20['Bat First'].isin(const_teams)) & (t20['Bat Second'].isin(const_teams))]
print(f'After Removing Irrelevant Columns : {t20.shape}')
print(f"Consistent Teams : \n{t20['Bat First'].unique()}")

t20.head()
```

```python
print(f'Before Removing Overs : {t20.shape}')

t20 = t20[t20['overs'] >= 5.0]
print(f'After Removing Overs : {t20.shape}')

t20.head()
```

```python
from seaborn import heatmap

numerical_t20_df = t20.select_dtypes(include=['int64', 'float64'])
heatmap(data=numerical_t20_df.corr(), annot=True)
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

le = LabelEncoder()
for col in ['Bat First', 'Bat Second']:
  t20[col] = le.fit_transform(t20[col])

t20.head()
```

```python
from sklearn.compose import ColumnTransformer

columnTransformer = ColumnTransformer([('encoder', OneHotEncoder(), [0, 1])], remainder='passthrough')
```

```python
t20_array = columnTransformer.fit_transform(t20)

# Convert the sparse matrix to an array
t20_array = t20_array.toarray()

# Create the list of column names for the DataFrame
feature_names = list(columnTransformer.named_transformers_['encoder'].get_feature_names_out(['Bat First', 'Bat Second'])) + ['Innings Runs', 'Innings Wickets', 'Target Score', 'overs', 'runs_last_5_overs', 'wickets_last_5_overs']
```

```python
df = pd.DataFrame(t20_array, columns=feature_names)
df.head()
```

```python
features = df.drop(['Target Score'], axis=1)
labels = df['Target Score']
```

```python
from sklearn.model_selection import train_test_split

train_features, test_features, train_labels, test_labels = train_test_split(features, labels, test_size=0.20, shuffle=True)
print(f"Training Set : {train_features.shape}\nTesting Set : {test_features.shape}")
```

```python
models = dict()
```

```python
from sklearn.tree import DecisionTreeRegressor
tree = DecisionTreeRegressor()
# Train Model
tree.fit(train_features, train_labels)
```

```python
# Evaluate Model
train_score_tree = str(tree.score(train_features, train_labels) * 100)
test_score_tree = str(tree.score(test_features, test_labels) * 100)
print(f'Train Score : {train_score_tree[:5]}%\nTest Score : {test_score_tree[:5]}%')
models["tree"] = test_score_tree
```

```python
from sklearn.metrics import mean_absolute_error as mae, mean_squared_error as mse
print("---- Decision Tree Regressor - Model Evaluation ----")
print("Mean Absolute Error (MAE): {}".format(mae(test_labels, tree.predict(test_features))))
print("Mean Squared Error (MSE): {}".format(mse(test_labels, tree.predict(test_features))))
print("Root Mean Squared Error (RMSE): {}".format(np.sqrt(mse(test_labels, tree.predict(test_features)))))
```

```python
from sklearn.ensemble import RandomForestRegressor
forest = RandomForestRegressor()
# Train Model
forest.fit(train_features, train_labels)
```

```python
# Evaluate Model
train_score_forest = str(forest.score(train_features, train_labels)*100)
test_score_forest = str(forest.score(test_features, test_labels)*100)
print(f'Train Score : {train_score_forest[:5]}%\nTest Score : {test_score_forest[:5]}%')
models["forest"] = test_score_forest
```

```python
print("---- Random Forest Regression - Model Evaluation ----")
print("Mean Absolute Error (MAE): {}".format(mae(test_labels, forest.predict(test_features))))
print("Mean Squared Error (MSE): {}".format(mse(test_labels, forest.predict(test_features))))
print("Root Mean Squared Error (RMSE): {}".format(np.sqrt(mse(test_labels, forest.predict(test_features)))))
```

```python
import matplotlib.pyplot as plt

model_names = list(models.keys())
accuracy = list(map(float, models.values()))

# creating the bar plot
plt.bar(model_names, accuracy)
```

```python
def score_predict(batting_team, bowling_team, runs, wickets, overs, runs_last_5_overs, wickets_last_5_overs, model=forest):
    # Build one-hot vectors programmatically to avoid long hard-coded branches
    teams = ['Afghanistan', 'Australia', 'Bangladesh', 'England', 'India', 'Ireland', 'New Zealand', 'Pakistan', 'South Africa', 'Sri Lanka', 'West Indies', 'Zimbabwe']

    def one_hot(team):
        return [1 if t == team else 0 for t in teams]
    batting_vec = one_hot(batting_team)
    bowling_vec = one_hot(bowling_team)

    feature_vec = batting_vec + bowling_vec + [runs, wickets, overs, runs_last_5_overs, wickets_last_5_overs]
    arr = np.array([feature_vec])

    pred = model.predict(arr)
    return int(round(pred[0]))
```

```python
batting_team='India'
bowling_team='South Africa'
score = score_predict(batting_team, bowling_team, runs=195, wickets=3, overs=17, runs_last_5_overs= 80, wickets_last_5_overs=0)
print(f'Predicted Score : {score}')
```
